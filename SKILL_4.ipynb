```python
import numpy as np
from tensorflow import keras
from tensorflow.keras.layers import Embedding, Conv1D, GlobalMaxPool1D, Dense, Dropout, Input
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
texts = [
    "I love this movie",
    "This film was great",
    "Amazing acting and story",
    "I hated the plot",
    "Terrible movie",
    "Bad acting",
    "I enjoyed it",
    "Not good"
]
labels = [1,1,1,0,0,0,1,0] # 1=positive,0=negative
```

```python
# ---- tokenize & pad ----
tokenizer = Tokenizer(oov_token="<unk>")
tokenizer.fit_on_texts(texts)
sequences = tokenizer.texts_to_sequences(texts)
maxlen = 6
X = pad_sequences(sequences, maxlen=maxlen, padding="post")
y = np.array(labels)

vocab_size = len(tokenizer.word_index) + 1  # include padding
```

```python
# ---- model ----
inp = Input(shape=(maxlen,))
x = Embedding(input_dim=vocab_size, output_dim=16, input_length=maxlen)(inp)
x = Conv1D(filters=32, kernel_size=3, activation="relu")(x)
x = GlobalMaxPool1D()(x)
x = Dropout(0.5)(x)
out = Dense(1, activation="sigmoid")(x)
model = Model(inp, out)
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
```

```python
#---- train ----
model.fit(X, y, epochs=20, batch_size=4, verbose=2)
```

```python
#---- predict example ----
test_sentences = ["I really enjoyed the film", "It was awful and boring"]
test_seq = tokenizer.texts_to_sequences(test_sentences)
test_X = pad_sequences(test_seq, maxlen=maxlen, padding="post")
probs = model.predict(test_X)
print("Predictions (prob positive):", probs.squeeze())
```
